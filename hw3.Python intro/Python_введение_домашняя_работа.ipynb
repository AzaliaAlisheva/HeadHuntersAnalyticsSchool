## Как оценивается работа

Максимум — 10 баллов. Они складываются из трёх частей.

| Что оценивается | Баллы | Как ставится |
|---|---|---|
| Правильность | 7 | по каждой задаче отдельно, разбалловка — в конце ноутбука |
| Читаемость кода | 1,5 | один раз за весь ноутбук |
| Стиль оформления | 1,5 | один раз за весь ноутбук |

### Правильность — 7 баллов

Балл за задачу ставится, если выполнены три условия:

- ноутбук выполняется сверху вниз без ошибок;
- результат совпадает с ожидаемым;
- соблюдены требования условия: имена переменных и функций, оговорки вроде «одним циклом» или «без `sum()`».

Если результат верный, но требование условия нарушено, за задачу ставится половина балла.

### Читаемость кода — 1,5 балла

Проверяющий читает ваш код так, как его читал бы коллега. Он смотрит на пять вещей:

- **Имена** объясняют, что лежит в переменной: `salary`, а не `s`.
- **Порядок.** Решение идёт по шагам: данные, расчёт, вывод. Лишних действий нет.
- **Чистота.** В ноутбуке не осталось закомментированных черновиков и отладочных `print`.
- **Повторы.** Одинаковые действия собраны в цикл или функцию, а не скопированы несколько раз.
- **Комментарии** объясняют смысл действия, а не пересказывают код. Если код понятен без них, они не нужны.

### Стиль оформления — 1,5 балла

Стиль оценивается по правилам PEP 8 из раздела 7.8 конспекта:

- имена переменных и функций — в стиле snake_case;
- отступ — 4 пробела на каждый уровень вложенности;
- пробелы стоят вокруг `=` и операторов и после запятых, но не перед скобкой вызова;
- между логическими частями кода — одна пустая строка, до и после определения функции — две.

### Шкала для читаемости и стиля

| Балл | Когда ставится |
|---|---|
| 1,5 | замечаний нет, или они единичные |
| 1 | одно и то же замечание повторяется в нескольких задачах |
| 0,5 | замечания есть почти в каждой задаче |
| 0 | код трудно читать |

Соискатели часто спрашивают, сколько они получат после вычета налога.

````python
# Ваше решение:
def net_salary(gross, tax_percent=13):
    """
    Возвращает зарплату «на руки»: оклад за вычетом налога

    Parameters:
        gross (int, float): оклад
        tax_percent (int, float): налог в процентах

    Returns:
        float: зарплата за вычетом налога
    """
    return gross * (1 - tax_percent / 100)

print(net_salary(100000))
print(net_salary(100000, 15))
````

### Задача 6.2. Количество и доля высоких зарплат

В отчёте нужна доля вакансий с зарплатой не ниже порога. Рядом с долей всегда показывают абсолютные числа, поэтому понадобятся две функции.

1. `count_above(salaries, threshold)` возвращает, сколько зарплат в списке не ниже порога `threshold`.
2. `share_above(salaries, threshold)` возвращает **долю** таких зарплат. Внутри она должна **вызывать** `count_above`. Для пустого списка функция возвращает `None`, а не останавливается с делением на ноль.

Выведите результат для списка `sample` и порога `120000` f-строкой — в том виде, как в ожидаемом результате. Затем выведите результат вызова `share_above([], 120000)`.

**Ожидаемый результат:**

```
Доля зарплат от 120000: 60.0% (3 из 5)
None
```

````python
sample = [90000, 150000, 120000, 210000, 60000]

# Ваше решение:
def count_above(salaries, threshold):
    return len([salary for salary in salaries if salary >= threshold])

def share_above(salaries, threshold):
    if not salaries:
        return None
    return count_above(salaries, threshold) / len(salaries) * 100.0

threshold = 120000
print(f"Доля зарплат от {threshold}: {share_above(sample, threshold):.1f}% ({\
    count_above(sample, threshold)} из {len(sample)})")
print(share_above([], 120000))
````

Напишите функцию `load_vacancies(path)`:

- она читает файл по пути `path` и **возвращает** список словарей с ключами `"name"`, `"city"`, `"salary"`, `"responses"` и `"invites"`;
- название и город — строки, остальные три значения — целые числа;
- пустые строки функция пропускает;
- если файла нет, она возвращает пустой список, а не останавливается с ошибкой.

Проверьте функцию так:

Руководитель просит короткий отчёт по вакансиям из списка `case_vacancies`. Напишите две функции: одна только считает, вторая только оформляет результат.

**1. `analyze_vacancies(vacancies)`** принимает список словарей и возвращает словарь-отчёт с ключами:

| Ключ | Значение |
|---|---|
| `"count"` | количество вакансий |
| `"responses"` | сумма откликов |
| `"invites"` | сумма приглашений |
| `"conversion"` | доля приглашений от откликов; `None`, если откликов нет |
| `"city_responses"` | словарь «город → сумма откликов» |
| `"top_salary_name"` | название вакансии с наибольшей зарплатой; `None`, если зарплата не указана ни в одной вакансии |

Требования к функции:

- она проходит по списку **одним циклом**;
- вакансии без зарплаты (`None`) в поиске наибольшей зарплаты не участвуют;
- для пустого списка функция не останавливается с ошибкой.

**2. `save_report(report, path)`** записывает отчёт в файл по пути `path` — ровно четыре строки, как в ожидаемом результате. Рядом с конверсией указаны абсолютные числа. Если конверсия равна `None`, последняя строка — `Конверсия: нет данных`.

Примените обе функции к `case_vacancies` и сохраните отчёт в файл `report.txt`. Затем выполните проверку:

```python
report = analyze_vacancies(case_vacancies)
save_report(report, "report.txt")

    print(file.read().strip())

```

**Ожидаемый результат:**

```
Вакансий: 40
Откликов: 11735
Приглашений: 1167
Конверсия: 9.9% (1167 из 11735)
{'Москва': 2820, 'Казань': 3595, 'Санкт-Петербург': 4280, 'Новосибирск': 1040}
Аналитик данных
{'count': 0, 'responses': 0, 'invites': 0, 'conversion': None, 'city_responses': {}, 'top_salary_name': None}
```

````python
# Ваше решение:
def analyze_vacancies(vacancies):
    report = {}
    report["count"] = len(vacancies)

    report["responses"] = 0
    report["invites"] = 0
    city_responses = {}
    top_salary_name = None
    top_salary = 0
    for vacancy in vacancies:
        report["responses"] += vacancy["responses"]
        report["invites"] += vacancy["invites"]

        if vacancy["city"] not in city_responses:
            city_responses[vacancy["city"]] = vacancy["responses"]
        else:
            city_responses[vacancy["city"]] += vacancy["responses"]

        if vacancy["salary"] is not None and top_salary < vacancy["salary"]:
            top_salary = vacancy["salary"]
            top_salary_name = vacancy["name"]

    report["conversion"] = report["invites"] / report["responses"] * 100.0 if report["responses"] else None
    report["city_responses"] = city_responses
    report["top_salary_name"] = top_salary_name

    return report


def save_report(report, path):
    with open(path, mode="w", encoding="utf-8") as file:
        file.write(f"Вакансий: {report["count"]}\n")
        file.write(f"Откликов: {report["responses"]}\n")
        file.write(f"Приглашений: {report["invites"]}\n")
        file.write(f"Конверсия: {f"{report["conversion"]:.1f}% ({report["invites"] \
        } из {report["responses"]})\n" if report["conversion"] is not None \
                                 else "нет данных\n"}")



report = analyze_vacancies(case_vacancies)
save_report(report, "report.txt")

with open("report.txt", encoding="utf-8") as file:
    print(file.read().strip())

print(report["city_responses"])
print(report["top_salary_name"])
print(analyze_vacancies([]))
````
